```python
print("AUTOMATIC POWER FACTOR CORRECTION SYSTEM")
print("----------------------------------------")
print("Project: APFC System")
print("Purpose: To improve power factor")
print("Method: Automatic Capacitor Bank Selection")
```

```python
# Initial Power Factor Calculation

active_power = 10       # kW
apparent_power = 14.28  # kVA

power_factor = active_power / apparent_power

print("Active Power:", active_power, "kW")
print("Apparent Power:", apparent_power, "kVA")
print("Initial Power Factor:", round(power_factor, 2))
```

```python
import math

# APFC Capacitor Bank Calculation

active_power = 10
initial_pf = 0.70
target_pf = 0.95

angle1 = math.acos(initial_pf)
angle2 = math.acos(target_pf)

capacitor_kvar = active_power * (
    math.tan(angle1) - math.tan(angle2)
)

print("Active Power:", active_power, "kW")
print("Initial Power Factor:", initial_pf)
print("Target Power Factor:", target_pf)
print("Required Capacitor Bank:",
      round(capacitor_kvar, 2), "kVAR")
```

```python
# Automatic Capacitor Bank Selection

required_kvar = 6.92

capacitor_steps = [2, 2, 2, 1]

selected_steps = []
total_kvar = 0

for capacitor in capacitor_steps:
    if total_kvar + capacitor <= required_kvar:
        selected_steps.append(capacitor)
        total_kvar += capacitor

print("Required Compensation:", required_kvar, "kVAR")
print("Selected Capacitor Steps:", selected_steps)
print("Total Capacitor Rating:", total_kvar, "kVAR")
```

```python
import math

# Improved Power Factor Calculation

active_power = 10
initial_pf = 0.70
compensation = 6

initial_reactive_power = active_power * math.tan(
    math.acos(initial_pf)
)

remaining_reactive_power = (
    initial_reactive_power - compensation
)

new_apparent_power = math.sqrt(
    active_power**2 + remaining_reactive_power**2
)

improved_pf = active_power / new_apparent_power

print("Initial Power Factor:", initial_pf)
print("Selected Compensation:", compensation, "kVAR")
print("Improved Power Factor:", round(improved_pf, 3))
```

```python
import matplotlib.pyplot as plt

power_factors = [0.70, 0.922]
labels = ["Before Compensation", "After Compensation"]

plt.bar(labels, power_factors)

plt.title("APFC System - Power Factor Improvement")
plt.ylabel("Power Factor")
plt.ylim(0, 1.1)

plt.show()
```

```python
import math

# Testing Different Load Conditions

loads = [
    (10, 0.70),
    (8, 0.75),
    (12, 0.80)
]

capacitor_kvar = 6

for power, pf in loads:

    reactive_power = power * math.tan(math.acos(pf))

    remaining_reactive = reactive_power - capacitor_kvar

    apparent_power = math.sqrt(
        power**2 + remaining_reactive**2
    )

    improved_pf = power / apparent_power

    print("Load:", power, "kW")
    print("Initial PF:", pf)
    print("Improved PF:", round(improved_pf, 3))
    print("-------------------------")
```

```python
import matplotlib.pyplot as plt
import numpy as np

loads = ["10 kW", "8 kW", "12 kW"]

initial_pf = [0.70, 0.75, 0.80]
improved_pf = [0.922, 0.991, 0.970]

x = np.arange(len(loads))
width = 0.35

plt.bar(x - width/2, initial_pf, width,
        label="Before Compensation")

plt.bar(x + width/2, improved_pf, width,
        label="After Compensation")

plt.xticks(x, loads)
plt.ylabel("Power Factor")
plt.xlabel("Load Conditions")
plt.title("APFC Performance Analysis")
plt.ylim(0, 1.1)
plt.legend()
plt.grid(axis="y")

plt.show()
```
